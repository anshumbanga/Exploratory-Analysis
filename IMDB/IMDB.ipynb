## Feature Selection & Data Observations

### 1) Variance Threshold (0.01 – 0.05)

* Columns with very low variance do not provide useful information.
* Features with variance close to 0 can be removed.

### 2) `is_adult`

* Contains only one unique value (0).
* Variance is 0, so the column can be dropped.

### 3) `year_of_release`

* Data ranges from 2001 to 2019.
* Covers 18 years of information.
* Can be used to analyze time-based trends.

### 4) `imdb_votes`

* Mean is around 5,000 and standard deviation is around 14,500.
* High standard deviation compared to mean indicates right-skewed distribution.
* Log transformation can help in handling skewness.

```python
# upper fence and lower fence 

q1 = df.imdb_votes.quantile(0.25)
med = df.imdb_votes.median()
q3 = df.imdb_votes.quantile(0.75)

iqr = q3-q1 

print('IQR\t\t\t',q3-q1)
uf = q3 + 1.5*iqr 
lf = q1- 1.5*iqr 
print('upper fence\t\t',uf)
print('lower fence\t\t',lf)
print('number of outliers\t',df[df.imdb_votes>uf].shape[0])
print('potential outliers')
df[df.imdb_votes>uf].sample(5)
```

## Missing Values Summary

* `poster_path` → 103 null values
* `story` → 20 null values
* `tagline` → 1072 null values
* `actors` → 5 null values
* `win_nominations` → 922 null values
* `release_date` → 107 null values

### Observations

* `tagline` and `win_nominations` have a high number of missing values and may require imputation or careful consideration before modeling.
* `poster_path` and `release_date` have moderate missing values and can be handled using imputation or removal based on use case.
* `story` and `actors` have very few nulls and can be cleaned easily (drop or fill).
